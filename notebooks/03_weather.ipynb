```python
import requests
import pandas as pd
```

```python
url = "https://archive-api.open-meteo.com/v1/archive"
```

```python
params = {
    "latitude": 43.65,
    "longitude": -79.38,
    "start_date": "2024-01-01",
    "end_date": "2024-12-31",
    "hourly": "temperature_2m,precipitation,rain,snowfall,wind_speed_10m",
    "timezone": "America/Toronto",
}

response = requests.get(url, params=params)
print(response.status_code)   
```

```python
data = response.json()
print(data.keys())            # see the top-level structure
print(data["hourly"].keys())  # the columns we asked for

weather = pd.DataFrame(data["hourly"])
weather["time"] = pd.to_datetime(weather["time"])
weather.head()
```

```python

print(weather.shape)          # expect 8784 rows (2024 is a leap year: 366 x 24)
print(weather.isna().sum())
print(weather.describe())
```

```python
print(weather.shape)          # expect 8784 rows (2024 is a leap year: 366 x 24)
print(weather.isna().sum())
print(weather.describe())
```

```python
weather.to_csv("../data/weather_2024.csv", index=False)
```
