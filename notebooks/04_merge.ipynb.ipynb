```python
import pandas as pd
```

```python
df = pd.read_excel(r"C:\Users\HP\Desktop\Data Science\Transit Delay and Commute Predictor\data\ttc-bus-delay-data-2024.xlsx")
```

```python
df["datetime"] = pd.to_datetime(
    df["Date"].dt.strftime("%Y-%m-%d") + " " + df["Time"].astype(str)
)
df["hour_ts"] = df["datetime"].dt.floor("h")

df = df.dropna(subset=["Route"])
df = df[df["Min Delay"] > 0]
print("Delay rows:", len(df))
```

```python
weather = pd.read_csv("../data/weather_2024.csv", parse_dates=["time"])
print(weather["time"].dtype, df["hour_ts"].dtype)
```

```python
merged = df.merge(weather, left_on="hour_ts", right_on="time", how="left")

print("Rows after merge:", len(merged))            # must equal 52815
print(merged["temperature_2m"].isna().sum())       # should be 0
merged.head()
```

```python
merged.to_csv("../data/merged_2024.csv", index=False)
```
